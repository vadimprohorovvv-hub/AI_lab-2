__Задание 3__ Объединение и разбиение массивов

```python
import numpy as np

arr1 = np.random.randint(0, 51, 5)
arr2 = np.random.randint(0, 51, 5)

print(arr1)
print(arr2)

arr_combined = np.vstack((arr1, arr2))
print(arr_combined)

arr_split1, arr_split2 = np.vsplit(arr_combined, 2)
print(arr_split1.flatten())
print(arr_split2.flatten())
```

__Задание 4__ Генерация и работа с линейными последовательностями

```python
import numpy as np

arr = np.linspace(-10, 10, 50)

total_sum = np.sum(arr)
positive_sum = np.sum(arr[arr > 0])
negative_sum = np.sum(arr[arr < 0])

print(total_sum)
print(positive_sum)
print(negative_sum)
```

__Задание 5__ Работа с диагональными и единичными матрицами

```python
import numpy as np

identity_matrix = np.eye(4)
diagonal_matrix = np.diag([5, 10, 15, 20])

sum_identity = np.sum(identity_matrix)
sum_diagonal = np.sum(diagonal_matrix)

print(identity_matrix)
print(diagonal_matrix)
print(sum_identity)
print(sum_diagonal)

if sum_identity > sum_diagonal:
    print("Сумма элементов единичной матрицы больше")
elif sum_identity < sum_diagonal:
    print("Сумма элементов диагональной матрицы больше")
else:
    print("Суммы равны")
```

__Задание 6__ Создание и базовые операции с матрицами

__Задание 8__ Определитель и обратная матрица

```python
import numpy as np

matrix = np.random.randint(1, 11, (3, 3))

print(matrix)

det = np.linalg.det(matrix)
print(det)

if det != 0:
    inv_matrix = np.linalg.inv(matrix)
    print(inv_matrix)
else:
    print("Матрица вырождена")
```

__Задание 10__ Системы линейных уравнений

__Задание 11__ Индивидульное задание.

__Условие:__ В диете наобходимо велючить 3 продукта: мясо, рыбу и овощи. Они содержат белки, жиры и углеводы в разном количестве.
        В 100 г __мяса__ содержится 20 г белков, 10 г жиров и 5 г углеводов
        В 100 г __рыбы__ 15 г белка, 5 г жиров, 10 г углеводов
        В 100 г __овощей__ 5 г белка, 2 г жиров, 150 г углеводов
        Сколько граммов каждого продукта нужно съесть?
        

```python
import numpy as np
from scipy.optimize import lsq_linear
A = np.array([[20, 15, 5],
              [10, 5, 2],
              [5, 10, 20]])
B = np.array([100, 50, 150])
bounds = (0, np.inf)
result = lsq_linear(A, B, bounds=bounds)
print("Решение (в сотнях граммов):", result.x)
print("Мясо: {:.1f} г".format(result.x[0] * 100))
print("Рыба: {:.1f} г".format(result.x[1] * 100))
print("Овощи: {:.1f} г".format(result.x[2] * 100))
```
